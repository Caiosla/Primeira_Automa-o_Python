# Automação de Sistemas e Processos com Python

### Desafio:

Todos os dias, o sistema da empresa atualiza as vendas do dia anterior.
O trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior.

E-mail da diretoria: `diretoria@empresa.com`<br>
Local onde o sistema disponibiliza as vendas do dia anterior: `https://exemplo.com/sistema-de-vendas`

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado.

Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
# vamos precisar instalar
# pyautogui -> automações
# pandas -> dados
# openpyxl -> trabalhar com bases de dados em Excel (.xlsx)
# para instalar: abrir o terminal e rodar: pip install pandas pyautogui openpyxl pyperclip

# pyautogui.click -> clicar com o mouse
# pyautogui.write -> escrever um texto
# pyautogui.press -> aperta 1 tecla
# pyautogui.hotkey -> combinação de teclas
```

```python
import pyautogui
import time

pyautogui.PAUSE = 0.5

# Passo a passo do desafio
# Passo 1: Entrar no sistema da empresa (link do sistema)
# abrir o Chrome
pyautogui.press("win")
pyautogui.write("chrome")
pyautogui.press("enter")

time.sleep(2)

# digitar o link do sistema
link = "https://exemplo.com/sistema-de-vendas"
pyautogui.hotkey("ctrl", "l")
pyautogui.write(link)
# apertar enter e esperar
pyautogui.press("enter")
time.sleep(6)
```

```python
# Passo 2: Navegar no sistema para encontrar a base de dados
# OBS: as coordenadas abaixo são específicas da resolução/layout de tela
# de quem gravou o script original — remapeie antes de usar (veja a
# célula final "descobrir posição do mouse").
pyautogui.click(x=495, y=523, clicks=2)
time.sleep(2)

# Passo 3: Exportar a base de dados (baixar o arquivo)
pyautogui.click(x=1790, y=527)  # clica nos 3 pontos
pyautogui.click(x=1571, y=566)  # clica no baixar
pyautogui.press("enter")  # baixa o arquivo
time.sleep(3)
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pandas

# Passo 4: Calcular os indicadores (faturamento e quantidade de produtos vendidos)
# abrir a base de dados baixada
caminho = r"C:\Users\SEU_USUARIO\Downloads\vendas.xlsx"
tabela = pandas.read_excel(caminho)

# ver as informações da base de dados
display(tabela)

# somar o faturamento de todos os produtos = somar a coluna de Valor Final
faturamento = tabela["Valor Final"].sum()
# somar a quantidade de produtos = somar a coluna de quantidade
qtde_produtos = tabela["Quantidade"].sum()

print(faturamento)
print(qtde_produtos)
```

### Vamos agora enviar um e-mail pelo Gmail

```python
import pyperclip

# Passo 5: Enviar as informações por e-mail
# abrir uma nova aba
pyautogui.hotkey("ctrl", "t")

# entrar no email
pyautogui.write("http://mail.google.com/")
pyautogui.press("enter")
time.sleep(10)

# clicar no botão escrever
pyautogui.click(x=112, y=222)
time.sleep(8)

# para quem enviar o email
pyautogui.write("diretoria@empresa.com")
pyautogui.press("tab")  # seleciona o email
pyautogui.press("tab")  # passar para o campo do assunto

# qual o assunto
pyperclip.copy("Relatório de Vendas")
pyautogui.hotkey("ctrl", "v")
pyautogui.press("tab")

# qual o corpo do email
texto = f"""
Prezados,

Segue o relatório de vendas de hoje.

Faturamento: R${faturamento:,.2f}
Quantidade de produtos vendidos: {qtde_produtos:,}

Qualquer dúvida estou à disposição.
Atenciosamente.
"""
pyperclip.copy(texto)
pyautogui.hotkey("ctrl", "v")

# enviar o email
pyautogui.click(x=1200, y=984)
```
